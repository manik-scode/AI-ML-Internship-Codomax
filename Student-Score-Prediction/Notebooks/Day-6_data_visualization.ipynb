## Objective

The objective of this task is to visualize the Student Score dataset using Matplotlib. Data visualization helps identify trends, relationships, and patterns within the dataset before building machine learning models.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("../data/preprocess/clear_missing_data.csv")
df = pd.DataFrame(data)
df.head()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df["Hours_Studied"], df["Exam_Score"])

plt.title("Hours Studied vs Exam Score")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")

plt.show()
```

```python
average_score = df.groupby("Gender")["Exam_Score"].mean()

plt.figure(figsize=(6,5))

plt.bar(average_score.index, average_score.values)

plt.title("Average Exam Score by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Score")

plt.show()
```

```python
sorted_df = df.sort_values("Hours_Studied")

plt.figure(figsize=(8,5))

plt.plot(sorted_df["Hours_Studied"],
         sorted_df["Exam_Score"])

plt.title("Hours Studied vs Exam Score")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")

plt.show()
```

# Conclusion

Successfully created scatter, bar, and line charts using Matplotlib. These visualizations helped understand relationships and patterns in the Student Score dataset, providing useful insights before machine learning model development.
